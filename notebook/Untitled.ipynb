```python
df = df.drop_duplicates()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.histplot(df["charges"], kde=True)
plt.show()
```

```python
sns.histplot(df["bmi"], kde=True)
plt.show()
```

```python
sns.histplot(df["bmi"], kde=True)
plt.show()
```

```python
sns.boxplot(
    x="smoker",
    y="charges",
    data=df
)
plt.show()
sns.boxplot(
    x="sex",
    y="charges",
    data=df
)
plt.show()
sns.countplot(
    x="region",
    data=df
)
plt.show()
```

```python
df_encoded = pd.get_dummies(
    df,
    drop_first=True
)

plt.figure(figsize=(12,8))

sns.heatmap(
    df_encoded.corr(),
    annot=True,
    cmap="coolwarm"
)

plt.show()
```

```python
sns.pairplot(
    df,
    vars=["age","bmi","children","charges"]
)
plt.show()
```

```python
df.to_csv(
    "../data/insurance_cleaned.csv",
    index=False
)
```
